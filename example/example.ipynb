## Plots

Hey, could you generate some really brief simple code that would plot a sine wave in matplotlib? And we'll see if we can get this to show in the notebook.

Certainly! Below is a short Python script using Matplotlib to plot a sine wave. You can copy and paste this code into your Jupyter Notebook or any other Python environment where you have Matplotlib installed.

```python
import matplotlib.pyplot as plt
import numpy as np

# Generate time points from 0 to 2*pi with 1000 data points
t = np.linspace(0, 2 * np.pi, 1000)

# Compute the sine of each time point
y = np.sin(t)

# Create a plot
plt.plot(t, y)

# Add title and labels
plt.title('Sine Wave')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')

# Show the plot
plt.show()
```

This code will generate and display a sine wave.

````
import matplotlib.pyplot as plt
import numpy as np

# Generate time points from 0 to 2*pi with 1000 data points
t = np.linspace(0, 2 * np.pi, 1000)

# Compute the sine of each time point
y = np.sin(t)

# Create a plot
plt.plot(t, y)

# Add title and labels
plt.title('Sine Wave')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')

# Show the plot
plt.show()
````

## Progress Bars

We can only use the regular TQDM as if we were in a terminal because the notebook version requires IPy widgets that I'm not going to implement

````
from tqdm import tqdm
import time

for i in tqdm(range(1000)):
    time.sleep(0.01)  # Simulate some work being done
````

Notice how the notebook version isn't going to increment. So be aware. 

````
from tqdm.notebook import tqdm
import time

for i in tqdm(range(1000)):
    time.sleep(0.01)  # Simulate some work being done
````

...See it just sits there and does nothing. So *don't try to use the notebook version.*

Good news is, if you just use the auto version of TQDM, you get the regular text version:

````
from tqdm.auto import tqdm
import time

for i in tqdm(range(1000)):
    time.sleep(0.01)  # Simulate some work being done
````
