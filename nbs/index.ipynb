## How to use

Launch the server: 
```bash
boopiter [--port 8000]
```
If any other `boopiter` server is running, it will be killed in favor of the new one.

Point your browser at that URL.

You can also aunch and pre-load a notebook:
```
boopiter nbs/00_core.ipynb
```
